# MOUSE configuration 125: why the correction chain matters

> **Supplementary poster/testing notebook.** Original processing date: **2026-09-15**; MoDaCor version: **1.8.0**. This is not part of the core MOUSE correction example.

This compact entry point runs the shared poster comparison notebook for configuration 125. The analysis and visual styling are intentionally identical to configuration 166 so the resulting figures can be compared directly.

```python
from pathlib import Path
import json
import os

for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
    options = (
        candidate / "MOUSE_166_correction_eyecatcher.ipynb",
        candidate / "BAM" / "MOUSE" / "supplementary" / "poster_2026" / "MOUSE_166_correction_eyecatcher.ipynb",
    )
    shared_notebook = next((path for path in options if path.is_file()), None)
    if shared_notebook is not None:
        break
else:
    raise FileNotFoundError("Could not locate MOUSE_166_correction_eyecatcher.ipynb.")

previous_configuration = os.environ.get("MOUSE_CONFIGURATION")
os.environ["MOUSE_CONFIGURATION"] = "125"
try:
    shared = json.loads(shared_notebook.read_text(encoding="utf-8"))
    for cell_index, cell in enumerate(shared["cells"]):
        if cell.get("cell_type") == "code":
            source = "".join(cell.get("source", []))
            exec(compile(source, f"{shared_notebook.name}:cell-{cell_index}", "exec"), globals())
finally:
    if previous_configuration is None:
        os.environ.pop("MOUSE_CONFIGURATION", None)
    else:
        os.environ["MOUSE_CONFIGURATION"] = previous_configuration
```
